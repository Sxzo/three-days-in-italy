```python
import pandas as pd

pd.set_option('display.max_rows', 120)
pd.set_option('display.max_colwidth', 100)

df = pd.read_json('data/italy.json')
print(df.shape)
df.columns
```

```python
df.head()
```

```python
# One full record, top to bottom
df.iloc[0]
```

```python
# What's missing, by column
df.isna().sum().sort_values(ascending=False)
```

```python
df['type'].value_counts()
```

```python
df['city'].value_counts()
```

```python
df['neighborhood'].unique()
```

```python
df['price_range'].value_counts()
```

```python
# Every distinct hours string
for h in sorted(df['hours'].dropna().unique()):
    print(h)
```

```python
# Hours that don't look like the usual 'Days HH:MM-HH:MM' shape
df[df['hours'].notna() & ~df['hours'].str.contains(':')][['name', 'type', 'hours']]
```

```python
# Every seasonal note
for note in df['seasonal_notes'].dropna():
    print(note)
```

```python
df['tags'].explode().value_counts()
```

```python
# local_favorite vs local-favorite — the one inconsistent tag
df[df['tags'].apply(lambda tags: 'local_favorite' in tags)][['id', 'name', 'tags']]
```

```python
df['rating'].describe()
```

```python
df.nsmallest(5, 'rating')[['name', 'type', 'city', 'rating']]
```

```python
df['duration_minutes'].value_counts().sort_index()
```

```python
# Floats, not bools — the single null drags the whole column to float
df['booking_required'].value_counts(dropna=False)
```

```python
# Rows with no hours — mostly neighborhoods, viewpoints, and day trips
df[df['hours'].isna()][['id', 'name', 'type', 'city', 'duration_minutes']]
```

```python
# Rows with no duration
df[df['duration_minutes'].isna()][['id', 'name', 'type', 'city', 'hours']]
```

```python
# Same coordinates, different rows
df[df.duplicated(['latitude', 'longitude'], keep=False)].sort_values('latitude')[
    ['id', 'name', 'type', 'city', 'latitude', 'longitude']
]
```

```python
# How many places of each type, per city
pd.crosstab(df['city'], df['type'])
```

```python
# Total minutes of stuff to do per city — Rome and Florence are the only ones with 3 days in them
df.groupby('city')['duration_minutes'].agg(['count', 'sum', 'mean']).sort_values('sum', ascending=False)
```

```python
# Lat/lon range per city — Milan's is suspiciously wide
df.groupby('city')[['latitude', 'longitude']].agg(['min', 'max']).round(3)
```

```python
# Brera Antique Market is in Milan but its longitude is 11.19 (should be ~9.19)
df[df['city'] == 'Milan'][['id', 'name', 'neighborhood', 'latitude', 'longitude']]
```

```python
# Everything in one city
df[df['city'] == 'Rome'][['name', 'type', 'neighborhood', 'hours', 'duration_minutes', 'rating']]
```

```python
# Everything with a given tag
df[df['tags'].apply(lambda tags: 'hidden-gem' in tags)][['name', 'type', 'city', 'rating']]
```

```python
# Keyword search in the descriptions
df[df['description'].str.contains('rooftop|sunset', case=False)][['name', 'city', 'description']]
```

```python
# Read a few descriptions in full
for _, row in df.sample(5, random_state=1).iterrows():
    print(f"{row['name']} ({row['type']}, {row['city']})")
    print(row['description'])
    print()
```
