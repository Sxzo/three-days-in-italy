```python
import json
import pandas as pd

pd.set_option("display.max_rows", 120)
pd.set_option("display.max_colwidth", 80)
```

```python
df = pd.read_json("data/italy.json")
df.shape
```

```python
# what the rows actually look like
df.head()
```

```python
# dtypes and where data is missing
df.info()
df.isna().sum()
```

```python
# numeric spread: rating, duration, coordinates
df.describe()
```

```python
# categorical breakdowns
for col in ["type", "city", "region", "price_range", "booking_required"]:
    print(df[col].value_counts(dropna=False), "\n")
```

```python
# tags is a list per row, so explode before counting
df["tags"].explode().value_counts()
```

```python
# hours is free text and sometimes null or seasonal
df["hours"].value_counts(dropna=False).head(20)
```

```python
# filter and sort
df.query("city == 'Rome' and rating >= 4.5").sort_values("rating", ascending=False)[
    ["name", "type", "rating", "duration_minutes", "price_range"]
]
```

```python
# group to compare cities
df.groupby("city").agg(
    places=("id", "count"),
    avg_rating=("rating", "mean"),
    avg_minutes=("duration_minutes", "mean"),
).sort_values("places", ascending=False)
```
